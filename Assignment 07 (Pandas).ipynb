4) From the DataFrame of Q1, select (a) only the Name and Salary columns, (b) rows where Age is greater than 30, (c) rows where City is 'Delhi' and Salary is above 40000.

6) Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

```python
import pandas as pd
data = {
    'Name': ['Ali', 'Ahmed', 'Sara', 'Zain', 'Bilal', 'Usman'],
    'Age': [25, 32, 28, 35, 29, 40],
    'City': ['Karachi', 'Delhi', 'Lahore', 'Delhi', 'Peshawar', 'Quetta'],
    'Salary': [35000, 45000, 50000, 60000, 40000, 42000]
}

df = pd.DataFrame(data)
df['Bonus'] = df['Salary'] * 0.10
df['Total'] = df['Salary'] + df['Bonus']
df = df.rename(columns={'City': 'Location'})
df = df.drop(columns=['Bonus'])

print(df)
```

7) Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
import pandas as pd
import numpy as np
data = {
    'Name': ['Ali', 'Ahmed', 'Sara', 'Zain', 'Bilal', np.nan],
    'Age': [25, np.nan, 28, 35, 29, 40],
    'City': ['Karachi', 'Delhi', 'Lahore', np.nan, 'Peshawar', 'Quetta'],
    'Salary': [35000, 45000, np.nan, 60000, 40000, 42000]
}

df = pd.DataFrame(data)
missing_count = df.isnull().sum()
print("Missing values count:\n", missing_count)
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
df = df.dropna()

print("\nFinal DataFrame after filling and dropping:\n", df)
```

8) Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
import pandas as pd
data = {
    'Name': ['Ali', 'Ahmed', 'Sara', 'Zain', 'Bilal', 'Usman'],
    'Age': [25, 32, 28, 35, 29, 40],
    'City': ['Karachi', 'Delhi', 'Lahore', 'Delhi', 'Peshawar', 'Quetta'],
    'Salary': [35000, 45000, 50000, 60000, 40000, 42000]
}

df = pd.DataFrame(data)
sorted_df = df.sort_values(by=['Salary', 'Age'], ascending=[False, True])
print("Sorted DataFrame:\n", sorted_df)
top_3_paid = df.nlargest(3, 'Salary')
print("\nTop 3 highest-paid employees:\n", top_3_paid)
```

9) Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
import pandas as pd
data = {
    'Name': ['Ali', 'Ahmed', 'Sara', 'Zain', 'Bilal', 'Usman'],
    'Age': [25, 32, 28, 35, 29, 40],
    'City': ['Karachi', 'Delhi', 'Lahore', 'Delhi', 'Peshawar', 'Quetta'],
    'Salary': [35000, 45000, 50000, 60000, 40000, 42000]
}

df = pd.DataFrame(data)
result = df.groupby('City')['Salary'].agg(['mean', 'count', 'max'])

print(result)
```

10) Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
import pandas as pd
emp_data = {
    'EmpID': [1, 2, 3, 4],
    'Name': ['Ali', 'Ahmed', 'Sara', 'Zain'],
    'DeptID': [101, 102, 101, 103]
}
df_emp = pd.DataFrame(emp_data)

dept_data = {
    'DeptID': [101, 102, 104],
    'DeptName': ['HR', 'IT', 'Finance']
}
df_dept = pd.DataFrame(dept_data)

inner_join_df = pd.merge(df_emp, df_dept, on='DeptID', how='inner')
print("--- Inner Join ---\n", inner_join_df)
left_join_df = pd.merge(df_emp, df_dept, on='DeptID', how='left')
print("\n--- Left Join ---\n", left_join_df)
df1 = pd.DataFrame({'Name': ['Ali', 'Sara'], 'Salary': [35000, 50000]})
df2 = pd.DataFrame({'Name': ['Sara', 'Zain'], 'Salary': [50000, 60000]})

concatenated_df = pd.concat([df1, df2]).drop_duplicates()
print("\n--- Concat and Drop Duplicates ---\n", concatenated_df)
```
